### Kompletter Workflow um Daten aus der Datenbank zu holen für Regression

```python
import pandas as pd
import psycopg2

# Verbindung aufbauen
conn = psycopg2.connect(
    host="localhost",
    database="berlin_db",
    user="postgres",
    password="dein_passwort"
)

# Daten direkt in Pandas laden
df = pd.read_sql('SELECT price, minimum_nights, reviews_per_month FROM "Test_listing_import";', conn)

# Verbindung schließen
conn.close()

# Daten vorbereiten
X = df[["price", "minimum_nights"]]
y = df["reviews_per_month"]

# Regression
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)
```

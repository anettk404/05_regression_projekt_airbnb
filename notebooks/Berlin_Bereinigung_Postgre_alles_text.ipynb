1. Versuch: Hier werden die Daten von listings_Berlin.csv bereinigt um sie in die Datenbank mit pg admin zu laden. 

```python
import pandas as pd

# CSV einlesen
df = pd.read_csv("listings_Berlin.csv")

# Alle Spalten als string (object)
df = df.astype(str)

# price: nur Zahlen behalten, sonst leeren String
df['price'] = df['price'].str.extract('(\d+)')[0]

# NaN in leeren Strings umwandeln
df = df.fillna('')

# Bereinigte CSV speichern
df.to_csv("listings_clean.csv", index=False)
```

Schwächen: Alle Spalten werden in Strings umgewandelt, was zu Datenverlust führen kann und für die Regression unpraktisch ist.

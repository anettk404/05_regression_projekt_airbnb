Ziel ist die Planung und der Aufbau einer IT-Infrastruktur für LehrerInnen. Dazu wurden Unterrichtsmaterialien der letzten Jahre (PDFs, Word-, PPT- und Excel-Dokumente, Videos, Audios, Moodle- und Goodnotes-Sicherungen, ...) auf einer externen Festplatte gesichtert und ein Raspberry Pi als Server eingerichtet, Docker installiert und MongoDB als Datenbank konfiguriert. 

Die Vision: Alle Unterrichtsmaterialien werden in einen "Data Lake" gesammelt und mit Tags versehen, um eine einfache Suche und Kategorisierung zu ermöglichen. Daten, die besonders geschützt werden müssen (Noten, Protokolle Elterngespräche, ...), werden verschlüsselt gespeichert. Lehrkräfte können über eine Weboberfläche auf die Materialien zugreifen (Nextcloud), sie durchsuchen und neue Materialien hinzufügen. Weitere Personen (KollegInnen oder SchülerInnen) können zeitweise Zugriff auf den Server erhalten, um gemeinsam an den Materialien zu arbeiten.

In diesem Notebook wird die Verbindung zwischen dem Raspberry Pi und dem lokalen Computer hergestellt.

```python
from pymongo import MongoClient
import sys

# SETZE HIER DIE IP DEINES PI EIN
PI_IP = "193.210.220.80" 

def test_connection():
    print(f"Versuche Verbindung zu {PI_IP} aufzubauen...")
    
    # timeout von 2 Sekunden, damit wir nicht ewig warten
    client = MongoClient(f"mongodb://{PI_IP}:27017/", serverSelectionTimeoutMS=2000)
    
    try:
        # Der 'ping' Befehl prüft, ob die DB antwortet
        client.admin.command('ping')
        print("✅ ERFOLG! Der Raspberry Pi hat geantwortet.")
        print("Deine Teachercloud-Datenbank ist online.")
        
    except Exception as e:
        print(f"❌ FEHLER: Verbindung fehlgeschlagen.")
        print(f"Details: {e}")
        print("\nCheckliste:")
        print(f"1. Ist die IP {PI_IP} korrekt?")
        print("2. Zeigt 'docker ps' auf dem Pi '0.0.0.0:27017->27017/tcp'?")
        print("3. Sind Laptop und Pi im selben WLAN?")

if __name__ == "__main__":
    test_connection()
```
